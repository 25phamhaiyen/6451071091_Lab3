```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu a) Đọc bộ dữ liệu Iris từ sklearn
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc bộ dữ liệu Iris
iris = load_iris()
print("Đã đọc thành công bộ dữ liệu Iris từ thư viện sklearn!")
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu b) Khảo sát bộ dữ liệu Iris
print("--- 1. Tên các thuộc tính và thuộc tính phân lớp ---")
print("Tên thuộc tính (feature_names):", iris.feature_names)
print("Thuộc tính phân lớp (target_names):", iris.target_names)

print("\n--- 2. Số lượng mẫu và số lượng thuộc tính ---")
print("Kích thước dữ liệu (data.shape):", iris.data.shape)
print("Kích thước nhãn (target.shape):", iris.target.shape)
print(f"-> Bộ dữ liệu có {iris.data.shape[0]} mẫu và {iris.data.shape[1]} thuộc tính.")

print("\n--- 3. 5 mẫu dữ liệu đầu tiên ---")
for i in range(5):
    print(f"Mẫu {i+1}: Đặc trưng = {iris.data[i]} | Nhãn = {iris.target[i]} ({iris.target_names[iris.target[i]]})")

print("\n--- 4. Thống kê count, mean, std, min, max, tứ phân vị bằng DataFrame describe() ---")
df_iris = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df_iris['target'] = iris.target
df_iris['species'] = df_iris['target'].map(lambda x: iris.target_names[x])
display(df_iris.describe())
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu c) Chuẩn hóa dữ liệu data về đoạn [0, 1]
# Gán X là data và Y là target của bộ dữ liệu
X = iris.data.copy()
Y = iris.target.copy()

# Chia mỗi giá trị của X cho giá trị lớn nhất trong cột tương ứng
max_features = np.max(X, axis=0)
X = X / max_features

print("Giá trị lớn nhất của từng cột thuộc tính:", max_features)
print("5 dòng đầu tiên của ma trận X sau khi chuẩn hóa:")
print(np.round(X[:5], 4))
print("Giá trị Min từng cột sau chuẩn hóa:", np.min(X, axis=0))
print("Giá trị Max từng cột sau chuẩn hóa:", np.max(X, axis=0))
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu d) Chia dữ liệu thành tập train và tập test theo tỷ lệ 8:2
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42, stratify=Y
)

print(f"Kích thước tập train: X_train = {X_train.shape}, y_train = {y_train.shape}")
print(f"Kích thước tập test:  X_test  = {X_test.shape}, y_test  = {y_test.shape}")
print(f"Tỷ lệ: Train = {len(X_train)/len(X)*100:.1f}%, Test = {len(X_test)/len(X)*100:.1f}%")
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu e) Sử dụng Cây quyết định (Decision Tree) để phân lớp
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Huấn luyện mô hình
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# Dự đoán nhãn lớp cho tập test
y_pred_dt = dt_model.predict(X_test)

# Tính và in Accuracy
acc_dt = accuracy_score(y_test, y_pred_dt)
print(f"Accuracy của Decision Tree trên tập test: {acc_dt * 100:.2f}%\n")

# In Precision, Recall, F1-score
print("--- Classification Report (Decision Tree) ---")
print(classification_report(y_test, y_pred_dt, target_names=iris.target_names))

# Hiển thị confusion matrix bằng heat map
cm_dt = confusion_matrix(y_test, y_pred_dt)
plt.figure(figsize=(6, 4.5))
sns.heatmap(cm_dt, annot=True, fmt='d', cmap='Blues', 
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.title(f"Confusion Matrix - Decision Tree (Accuracy: {acc_dt*100:.2f}%)")
plt.xlabel("Dự đoán (Predicted)")
plt.ylabel("Thực tế (Actual)")
plt.show()
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu f) Sử dụng Naïve Bayes để phân lớp
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Huấn luyện mô hình
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

# Dự đoán nhãn lớp cho tập test
y_pred_nb = nb_model.predict(X_test)

# Tính và in Accuracy
acc_nb = accuracy_score(y_test, y_pred_nb)
print(f"Accuracy của Naïve Bayes trên tập test: {acc_nb * 100:.2f}%\n")

# In Precision, Recall, F1-score
print("--- Classification Report (Naïve Bayes) ---")
print(classification_report(y_test, y_pred_nb, target_names=iris.target_names))

# Hiển thị confusion matrix bằng heat map
cm_nb = confusion_matrix(y_test, y_pred_nb)
plt.figure(figsize=(6, 4.5))
sns.heatmap(cm_nb, annot=True, fmt='d', cmap='Greens', 
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.title(f"Confusion Matrix - Naïve Bayes (Accuracy: {acc_nb*100:.2f}%)")
plt.xlabel("Dự đoán (Predicted)")
plt.ylabel("Thực tế (Actual)")
plt.show()
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu g) Sử dụng K Nearest Neighbors Classifier để phân lớp
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Huấn luyện mô hình KNN
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

# Dự đoán nhãn lớp cho tập test
y_pred_knn = knn_model.predict(X_test)

# Tính và in Accuracy
acc_knn = accuracy_score(y_test, y_pred_knn)
print(f"Accuracy của KNN (k=5) trên tập test: {acc_knn * 100:.2f}%\n")

# In Precision, Recall, F1-score
print("--- Classification Report (K Nearest Neighbors) ---")
print(classification_report(y_test, y_pred_knn, target_names=iris.target_names))

# Hiển thị confusion matrix bằng heat map
cm_knn = confusion_matrix(y_test, y_pred_knn)
plt.figure(figsize=(6, 4.5))
sns.heatmap(cm_knn, annot=True, fmt='d', cmap='Oranges', 
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.title(f"Confusion Matrix - KNN (Accuracy: {acc_knn*100:.2f}%)")
plt.xlabel("Dự đoán (Predicted)")
plt.ylabel("Thực tế (Actual)")
plt.show()
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu h) Sử dụng Neural Network để phân lớp
import os
os.environ['KERAS_BACKEND'] = 'torch'
import keras
from keras.models import Sequential
from keras.layers import Dense, Input
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Xây dựng Neural Network gồm 02 hidden layers và 01 output layer
nn_model = Sequential([
    Input(shape=(4,)),
    Dense(10, activation='relu'),
    Dense(20, activation='relu'),
    Dense(3, activation='softmax')
])

# Biên dịch mô hình
nn_model.compile(optimizer='adam', 
                 loss='sparse_categorical_crossentropy', 
                 metrics=['accuracy'])

# Huấn luyện mô hình
print("Bắt đầu huấn luyện Neural Network (epochs=100, batch_size=1)...")
history = nn_model.fit(X_train, y_train, epochs=100, batch_size=1, verbose=1)

# Dự đoán nhãn lớp cho tập test
y_prob_nn = nn_model.predict(X_test)
y_pred_nn = np.argmax(y_prob_nn, axis=1)

# Tính và in Accuracy
acc_nn = accuracy_score(y_test, y_pred_nn)
print(f"\nAccuracy của Neural Network trên tập test: {acc_nn * 100:.2f}%\n")

# In Precision, Recall, F1-score
print("--- Classification Report (Neural Network) ---")
print(classification_report(y_test, y_pred_nn, target_names=iris.target_names))

# Hiển thị confusion matrix bằng heat map
cm_nn = confusion_matrix(y_test, y_pred_nn)
plt.figure(figsize=(6, 4.5))
sns.heatmap(cm_nn, annot=True, fmt='d', cmap='Purples', 
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.title(f"Confusion Matrix - Neural Network (Accuracy: {acc_nn*100:.2f}%)")
plt.xlabel("Dự đoán (Predicted)")
plt.ylabel("Thực tế (Actual)")
plt.show()
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu i) Sử dụng các mô hình phân lớp đã huấn luyện cho 03 mẫu dữ liệu mới
new_samples = np.array([
    [6.2, 2.9, 4.3, 1.3],
    [5.1, 3.5, 1.4, 0.2],
    [7.3, 2.8, 6.4, 2.1]
])

# Chuẩn hóa 3 mẫu mới theo max_features đã tính ở câu c
new_samples_scaled = new_samples / max_features

# Dự đoán với cả 4 mô hình
pred_dt = dt_model.predict(new_samples_scaled)
pred_nb = nb_model.predict(new_samples_scaled)
pred_knn = knn_model.predict(new_samples_scaled)
pred_nn = np.argmax(nn_model.predict(new_samples_scaled), axis=1)

print("--- KẾT QUẢ DỰ ĐOÁN CHO 03 MẪU MỚI ---")
for i, sample in enumerate(new_samples):
    print(f"\nMẫu {i+1}: Đặc trưng gốc = {sample}")
    print(f"  + Decision Tree:  Nhãn = {pred_dt[i]} ({iris.target_names[pred_dt[i]]})")
    print(f"  + Naïve Bayes:    Nhãn = {pred_nb[i]} ({iris.target_names[pred_nb[i]]})")
    print(f"  + KNN (k=5):      Nhãn = {pred_knn[i]} ({iris.target_names[pred_knn[i]]})")
    print(f"  + Neural Network: Nhãn = {pred_nn[i]} ({iris.target_names[pred_nn[i]]})")
```
