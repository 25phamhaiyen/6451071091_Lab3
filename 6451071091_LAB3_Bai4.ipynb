```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu a, b) Đọc và chuẩn hóa dữ liệu Iris về đoạn [0, 1]
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd

iris = load_iris()
X = iris.data.copy()
Y = iris.target.copy()

# Chuẩn hóa về [0, 1] bằng cách chia cho max từng cột
max_features = np.max(X, axis=0)
X = X / max_features

print("Giá trị max từng thuộc tính:", max_features)
print("Kích thước ma trận X sau chuẩn hóa:", X.shape)
print("5 dòng đầu tiên:")
print(np.round(X[:5], 4))
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu c) Định nghĩa hàm create_model
import os
os.environ['KERAS_BACKEND'] = 'torch'
import keras
from keras.models import Sequential
from keras.layers import Dense, Input

def create_model():
    model = Sequential([
        Input(shape=(4,)),
        Dense(10, activation='relu'),
        Dense(20, activation='relu'),
        Dense(3, activation='softmax')
    ])
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

print("Đã định nghĩa hàm create_model() thành công!")
model_sample = create_model()
model_sample.summary()
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu d) Huấn luyện mô hình bằng k-fold cross validation (với k=10)
from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

kfold = KFold(n_splits=10, shuffle=True, random_state=42)

models = []
acc_list = []
precision_list = []
recall_list = []
f1_list = []
class_metrics_list = []

fold = 1
for train_idx, val_idx in kfold.split(X):
    X_train_f, X_val_f = X[train_idx], X[val_idx]
    y_train_f, y_val_f = Y[train_idx], Y[val_idx]
    
    model = create_model()
    # Huấn luyện mô hình với epochs=100, batch_size=1
    model.fit(X_train_f, y_train_f, epochs=100, batch_size=1, verbose=0)
    
    # Dự đoán
    y_pred_prob = model.predict(X_val_f, verbose=0)
    y_pred_f = np.argmax(y_pred_prob, axis=1)
    
    # Tính các độ đo tổng thể (weighted average)
    acc = accuracy_score(y_val_f, y_pred_f)
    p, r, f1, _ = precision_recall_fscore_support(y_val_f, y_pred_f, average='weighted', zero_division=0)
    
    # Tính các độ đo theo từng class
    p_cls, r_cls, f1_cls, _ = precision_recall_fscore_support(y_val_f, y_pred_f, labels=[0, 1, 2], average=None, zero_division=0)
    class_metrics_list.append((p_cls, r_cls, f1_cls))
    
    models.append(model)
    acc_list.append(acc)
    precision_list.append(p)
    recall_list.append(r)
    f1_list.append(f1)
    
    print(f"Fold {fold:2d}: Accuracy = {acc*100:6.2f}% | Precision = {p*100:6.2f}% | Recall = {r*100:6.2f}% | F1-score = {f1*100:6.2f}%")
    fold += 1
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu e) In ra Average Accuracy, Average Recall, Average Precision, Average F1-score sau k-fold
avg_acc = np.mean(acc_list)
avg_prec = np.mean(precision_list)
avg_rec = np.mean(recall_list)
avg_f1 = np.mean(f1_list)

summary_df = pd.DataFrame({
    'Metric': ['Average Accuracy', 'Average Precision', 'Average Recall', 'Average F1-score'],
    'Giá trị (%)': [avg_acc * 100, avg_prec * 100, avg_rec * 100, avg_f1 * 100]
})
print("--- KẾT QUẢ TRUNG BÌNH QUA 10 FOLD ---")
display(summary_df)
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu f) In ra Average Recall, Average Precision, Average F1-score cho từng class sau k-fold
p_classes = np.mean([item[0] for item in class_metrics_list], axis=0)
r_classes = np.mean([item[1] for item in class_metrics_list], axis=0)
f1_classes = np.mean([item[2] for item in class_metrics_list], axis=0)

class_summary_df = pd.DataFrame({
    'Class': iris.target_names,
    'Average Precision (%)': p_classes * 100,
    'Average Recall (%)': r_classes * 100,
    'Average F1-score (%)': f1_classes * 100
})
print("--- KẾT QUẢ TRUNG BÌNH THEO TỪNG LỚP QUA 10 FOLD ---")
display(class_summary_df)
```

```python
print("Họ và tên: Phạm Hải Yến - MSSV: 6451071091 - Lớp: CQ.64.CNTT")
# Câu g) Sử dụng mô hình có F1-score tốt nhất đã huấn luyện ở trên cho 03 mẫu dữ liệu mới
best_idx = np.argmax(f1_list)
best_model = models[best_idx]
print(f"Mô hình tốt nhất là Fold {best_idx + 1} với F1-score = {f1_list[best_idx]*100:.2f}%\n")

new_samples = np.array([
    [6.2, 2.9, 4.3, 1.3],
    [5.1, 3.5, 1.4, 0.2],
    [7.3, 2.8, 6.4, 2.1]
])

# Chuẩn hóa 3 mẫu mới
new_samples_scaled = new_samples / max_features

# Dự đoán bằng mô hình tốt nhất
pred_prob = best_model.predict(new_samples_scaled, verbose=0)
pred_classes = np.argmax(pred_prob, axis=1)

print("--- KẾT QUẢ DỰ ĐOÁN 03 MẪU MỚI (BẰNG MÔ HÌNH TỐT NHẤT K-FOLD) ---")
for i, sample in enumerate(new_samples):
    cls_idx = pred_classes[i]
    cls_name = iris.target_names[cls_idx]
    confidence = pred_prob[i][cls_idx] * 100
    print(f"Mẫu {i+1}: Đặc trưng gốc = {sample}")
    print(f"  -> Dự đoán nhãn: {cls_idx} ({cls_name}) với độ tin cậy: {confidence:.2f}%")
```
